3. What is the difference between Normalization and Standardization?

Answer -
In data preprocessing, both techniques are used to scale features so they are on a similar scale, but they achieve this differently:

Normalization (Min-Max Scaling)

Normalization rescales the data into a fixed range, typically 0 to 1.

Formula: \(X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}\)

When to use: Use this when you know the distribution of your data does not follow a Gaussian (Normal) distribution, or when you need your data to be within a specific bounding box (e.g., for Image Processing or Neural Networks).

Sensitivity: It is highly sensitive to outliers. One extreme value can "squish" all other data points into a very small range.

Standardization (Z-Score Normalization)

Standardization rescales data so that it has a mean of 0 and a standard deviation of 1.

Formula: \(z = \frac{X - \mu}{\sigma}\) (where \(\mu \) is the mean and \(\sigma \) is the standard deviation).

When to use: Use this when your data follows a Gaussian distribution. It is preferred for algorithms that assume normally distributed data, such as Linear Regression, Logistic Regression, and Linear Discriminant Analysis.

Sensitivity: It is much less affected by outliers compared to normalization, as it does not have a bounding range.

4. A dataset has missing values in the “Age” column. Suggest two techniques to handle this and
explain when they should be used.

Answer -

Since "Age" is a numerical variable, here are two common techniques to handle missing values:

1. Mean/Median ImputationTechnique: Replace missing values with the average (mean) or the middle value (median) of the existing data in that column.

When to use: Use Mean if the data is normally distributed and lacks significant outliers. Use Median if the data is skewed or contains outliers, as it is a more robust measure of central tendency.

2. K-Nearest Neighbors (KNN) ImputationTechnique: Predict the missing age based on the ages of other "similar" individuals in the dataset (determined by other features like Income, Occupation, etc.).

When to use: Use this when the missingness is not completely random and you believe other features in the dataset can accurately help predict a person's age. It is more accurate than simple imputation but more computationally expensive.

7. Explain the difference between Data Integration and Data Mapping in ETL.

Answer -

While these terms are related, they represent different levels of the ETL process:

Data Integration: This is the broad process of combining data from many different sources (databases, files, APIs) into a single, unified view. It's the "big picture" goal of an ETL pipeline.

Data Mapping: This is a specific task within integration. It involves creating a "map" or blueprint that shows exactly which field in the source (e.g., cust_id) connects to which field in the target database (e.g., Customer_ID).

8. Explain why Z-score Standardization is preferred over Min-Max Scaling when outliers exist?

Answer - When your data contains outliers (extreme values), Z-score Standardization is generally better for the following reasons:


Min-Max Scaling Sensitivity: This method squishes all values into a tight range (like 0 to 1) based on the minimum and maximum values. If you have one massive outlier, it will "pull" the range so far that all your normal data points get compressed into a tiny, indistinguishable cluster near zero.


Z-score Robustness: Z-score scales data based on the mean and standard deviation. While outliers do affect the mean, they don't force the rest of the data into a tiny range. The "normal" data stays spread out and useful, while the outlier simply ends up with a very high Z-score (like +5.0).
